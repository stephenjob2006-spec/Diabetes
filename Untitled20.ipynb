```python
x=df.drop(['diabetes'],axis=1)
x.head()
```

```python
y=df['diabetes']
y
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=2)
```

```python
xtrain.shape,xtest.shape,ytrain.shape,ytest.shape
```

```python
scaler=MinMaxScaler()
xtrain_s=scaler.fit_transform(xtrain)
xtest_s=scaler.transform(xtest)
```

```python
knn= KNeighborsClassifier()
svc= SVC()
D=DecisionTreeClassifier()
rf= RandomForestClassifier()
ab=AdaBoostClassifier()
gb=GradientBoostingClassifier()
lr= LogisticRegression()
mn=MultinomialNB()
bn=BernoulliNB()
gn=GaussianNB()
xb= XGBClassifier()

lists=[knn,svc,D,rf,ab,gb,lr,mn,bn,gn,xb]
for i in lists:
    print('            ',i         )
    print('*********************************************')
    i.fit(xtrain,ytrain)
    ypredtest=i.predict(xtest)
    ypredtrain=i.predict(xtrain)
    print("train_accuracy",accuracy_score(ytrain,ypredtrain))
    print("test_accuracy",accuracy_score(ytest,ypredtest))
    print(ConfusionMatrixDisplay.from_predictions(ytest,ypredtest))
    plt.show()
    print()
```

```python
lists=[knn,svc,D,rf,ab,gb,lr,mn,bn,gn,xb]
for i in lists:
    print('                   ',i  )
    print('**********************************************')
    scores = cross_val_score(i,x,y,cv=3,scoring='accuracy',n_jobs=-1)
    print(f"CV ACCURACY Scores: {scores}")
    print(f"Mean Accuracy: {np.mean(scores)*100:2f}%\n")
```

```python
gbs=GradientBoostingClassifier()
gbs.fit(xtrain,ytrain)
ypredict=gbs.predict(xtest)
print(accuracy_score(ytest,ypredict))
```
